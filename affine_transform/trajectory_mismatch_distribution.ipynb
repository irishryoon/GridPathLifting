```python
import os, sys
import numpy as np
import matplotlib.pyplot as plt
import pickle

sys.path.append('.')
from trajectory import World, random_walk
from affine_transform.score_mismatch import score_mismatch
from affine_transform.get_transform_mat import get_transform_mat, apply_transform_mat
```

```python
world_size = (100, 100)
holes = [(30, 30, 70, 70)]
num_holes = len(holes)
world = World(world_size, holes)
```

```python
original_traj = pickle.load(open(f"trajectory/random_walk_{num_holes}holes.pkl", "rb"))[0]
plt.figure(figsize=(5, 5))
plt.plot(original_traj[:, 0], original_traj[:, 1])
```

```python
random_traj = []
num_traj = 100
for i in range(num_traj):
    random_traj.append(random_walk(25000, world, save=False, no_warnings=True))

pickle.dump(random_traj, open(f'affine_transform/random_walk_{num_holes}holes.pkl', 'wb'))    
```

```python
random_traj = pickle.load(open(f"affine_transform/random_walk_{num_holes}holes.pkl", "rb"))

for i in range(5):
    plt.figure(figsize=(3, 3))
    plt.plot(random_traj[i][:, 0], random_traj[i][:, 1])
```

```python
mismatch_scores = []
for i in range(len(random_traj)):
    aligned_random = apply_transform_mat(random_traj[i], get_transform_mat(original_traj, random_traj[i]))
    mismatch_scores.append(score_mismatch(original_traj, aligned_random))
```

```python
print(mismatch_scores)
np.mean(mismatch_scores), np.std(mismatch_scores)
```

```python
import seaborn as sns
plt.figure(figsize=(5, 5))
sns.histplot(mismatch_scores, bins=20)
```
